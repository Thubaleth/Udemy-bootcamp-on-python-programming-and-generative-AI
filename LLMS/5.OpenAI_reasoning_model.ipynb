```python
import os
from openai import OpenAI
from dotenv import load_dotenv
```

```python

load_dotenv()
api_key = os.getenv("OPENAI_API_KEY")
client = OpenAI(api_key=api_key)
```

```python
prompt = """ 
Write a bash script that reads a logfile where eachline is in format
'YYYY-MM-DD' HH:MM:SS - LEVEL -Message', and prints out the count of messages for each log level (e.g., INFO,ERROR,DEBUG).
"""
```

```python
response = client.chat.completions.create(
    model = 'o3-mini',
    reasoning_effort='medium', #reasoning models are for complex prompts like sceince,math etc
    messages=[
        {'role':'user', 'content':prompt}
    ]
)

print(response.choices[0].message.content)
```
